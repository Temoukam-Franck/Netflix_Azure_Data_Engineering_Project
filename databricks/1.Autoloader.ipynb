# Incremental Data Loading using AutoLoader

```python
checkpoint_Location = "abfss://silver@netfixstorageacc.dfs.core.windows.net/checkpoint"
```

```python
df = spark.readStream.format("cloudFiles")\
      .option("cloudFiles.format", "csv")\
      .option("cloudFiles.schemaLocation", checkpoint_Location)\
      .load("abfss://raw@netfixstorageacc.dfs.core.windows.net")
```

```python
display(df, checkpointLocation = checkpoint_Location)
```

```python
df.writeStream\
  .option("checkpointLocation", checkpoint_Location)\
  .trigger(processingTime = "10 seconds")\
  .start("abfss://bronze@netfixstorageacc.dfs.core.windows.net/netflix_titles.csv")
```
